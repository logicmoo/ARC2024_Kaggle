# ARC3 Debugger Runner

Interactive wrapper around the real `arc_agi.Arcade` environment.

```python
from pathlib import Path
import sys

PROJECT_ROOT = Path.cwd().parent
sys.path.insert(0, str(PROJECT_ROOT / 'python'))

from arc3_runner import Arc3Runner
from swipl_bridge import SWIPrologBridge
```

```python
runner = Arc3Runner('ls20', render_mode='terminal')
runner.action_table()
```

```python
# Simple action
obs = runner.step('ACTION1')
print('state:', runner.state_name())
print('history:', runner.history())
```

```python
# Prolog chooses an action from the current snapshot
prolog = SWIPrologBridge(PROJECT_ROOT / 'prolog' / 'arc3_agent.pl')
decision = prolog.choose_action(runner.summary_for_prolog())
decision
```

```python
# Execute the Prolog decision
runner.step(decision['action'], data=decision.get('data') or {})
```
