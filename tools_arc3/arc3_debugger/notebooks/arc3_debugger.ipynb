# ARC3 Turtle Debugger

Python notebook calling SWI-Prolog and displaying the ANSI debugger console.

```python
from pathlib import Path
import sys

PROJECT_ROOT = Path.cwd().parent
sys.path.insert(0, str(PROJECT_ROOT / 'python'))

from swipl_bridge import SWIPrologBridge
from ansi_console import display_ansi
```

```python
bridge = SWIPrologBridge(
    PROJECT_ROOT / 'prolog' / 'arc3_debugger.pl'
)

output = bridge.debug_program([
    'pendown',
    'set_cell',
    'fwd(4)',
    'rot(90)',
    'fwd(3)',
])

display_ansi(output)
```
